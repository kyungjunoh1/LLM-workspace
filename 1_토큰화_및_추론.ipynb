### LLM(Large Language Model)
- 사람의 언어를 이해하고 생성하는 AI
- 방대한 양의 텍스트 데이터를 학습한 대형 언어 모델
#### LLM 동작
1. 사용자 질문 이해
2. 학습했던 데이터 기반으로
3. 가장 자연스럽고 맞는 답을 생성
  - 확률적으로 답을 찾는 모델
  - "나는 밥을 먹었다 그리고 ___"
    - 확률 : 잤다(20%), 공부했다(10%), 운동했다(30%)

# 토큰화
### 문장을 작은 단위(token)로 쪼개는 과정
- 예) 나는 학교에 간다(공백기준) -> [나는, 학교에, 간다]
### 토큰화를 하는 이유
- 모델은 문자를 그대로 이해할 수 없고 숫자(벡터)로만 계산
- 문장을 토큰 단위로 나눠서 id를 부여해서 사용
  - 예) [나는, 학교에, 간다] -> [101, 102, 103]
# 패딩
### 길이가 다른 문장들의 길이를 서로 맞추는 작업
### 사용 이유
- 모델은 고정된 길이의 입력만 처리 가능
- 문장의 길이를 맞춰 패딩을 사용하면 배치 연산으로 빠르게 계산할 수 있다
# 결론
### - 토큰화 및 패딩을 주는 주된 이유는 효율적인 학습과 연산을 하기 위한 용도
### - 모델은 입력 길이가 모두 같아야 행렬(Matrix)로 묶어서 한번에 계산할 수 있다

```python
input_test = "나는 항상 맛있는 음식을 먹는다."
```

```python
input_token = input_test.split()
input_token
```

### 임베딩(embedding)
- 단어(토큰)를 숫자 벡터로 변환

```python
for i in range(len(input_token)):
  print(i,":",input_token[i])
```

# 허깅페이스 트랜스포머
### 허깅페이스(Hugging Face)
- 인공지능 모델, 특히 자연어 처리(NLP)모델을 쉽게 사용하도록 도와주는 사이트+라이브러리
  - 오픈소스 모델 저장소 역할을 하며, 전 세계 연구자들이 만든 모델을 공유하여 사용
  - python 라이브러리로 transformers를 제공
### 트랜스포머(Transformers)
- transformers는 Hugging Face에서 만든 파이썬 패키지
- 주요기능
  - tokenizer : 문장을 숫자로 바꾸는 도구(토큰화)

#### klue/roberta-base ( Encoder )
- klue/roberta-base 는 한국어 자연어 처리(NLP) 태스크를 위해 사전학습(Pretrained) 된 로버타(RoBERTa) 기반 한국어 모델
- 단순히 대규모 한국어 말뭉치, 문장 구조, 단어 의미, 문맥 관계 이해 정도 학습되어 있는 모델
- 사전 학습 목적
  - 대화 생성용 LLM과 달리 자연어 이해에 최적화됨
  - 질문 응답, 감정 분석, 문장 분류 등에 사용

```python
from transformers import AutoTokenizer
model_id = "klue/roberta-base"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
tokenizer("안녕하세요")
```

```python
tokenizer("안녕하세요", padding="max_length", max_length=8)
```

```python
data = ["나는","배가 부르다"]
token = tokenizer(data, padding="longest")
token
```

```python
print("기본 크기 : ", tokenizer.model_max_length)
tokenizer("안녕하세요", padding="max_length")
```

```python
token = tokenizer("안녕하세요", padding="max_length", truncation=True, max_length=8)
token
```

```python
tokenizer.convert_ids_to_tokens(token["input_ids"])
```

## 모델 구분
#### GPT 계열( Decoder )
- Decoder : 이전 토큰을 보고 다음 토큰 생성. 자연어 생성에 최적화(챗봇, 글쓰기 보조, 코드 생성 등)
  - decoder : 문장 생성
- 모델 종류 : GPT2, KoGPT2, Yi-Ko 등
#### T5 ( 혼합형 모델 )
- Encoder-Decoder 구조
  - bert + gpt = 둘다 가능(문장 이해 및 생성)
- 입력 이해 + 출력 생성 가능
- 번역, 요약, 질의응답 등

# 카카오 생성형 모델(LLaMA)
### LLaMA(Large Language Model Meta AI)
- 대규모 언어 모델
- 입력 문장(prompt)을 받아 다음 단어를 생성하는 방식
- 기본적으로 텍스트 생성용 모델(GPT 계열)
### 생성형 모델을 활용하여 챗봇
  - model_id = "kakaocorp/kanana-nano-2.1b-base" (생성형 언어모델(카카오) - 상식, 요약, 번역, 생활 등)
  - AutoModelForCausalLM.from_pretrained( model_id ) (데이터 생성이므로 정답이 없음)
  - 일반적인 질문 확인
---
### 단위
- B(Billion(10억))
- 2.1b : 21억 개 파라미터
  - 파라미터가 많을 수록 학습된 지식이 많다
  - 가볍고 빠른 모델, 추론 약함, 양자화 취약
- 7B
  - 균형 좋음, 추론 가능, 양자화 가능

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
model_id = "kakaocorp/kanana-nano-2.1b-base"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")
```

```python
import torch
torch.cuda.is_available()
```

```python
model.parameters()
```

```python
next( model.parameters() ).device
```

```python
tokenizer.model_max_length
```

```python
tokenizer("안녕하세요")
```

```python
tokenizer.pad_token = tokenizer.eos_token
```

```python
data = ["나는", "배가 부르다"]
token = tokenizer(data, padding=True)
token
```

```python
prompt = "삼성전자에 대해 설명해줘"
```

```python
tokenizer(prompt)
```

```python
inputs = tokenizer(prompt, return_tensors="pt")
inputs
```

```python
for k, v in inputs.items():
  print(k, " : ", v.device)
  #break;
```

```python
outputs = model.generate(
    input_ids=inputs['input_ids'],
    attention_mask=inputs['attention_mask']
)
```

```python
outputs
```

```python
tokenizer.decode(outputs[0])
```

```python
tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
#key : value, key:value
outputs = model.generate(
    **inputs,
    max_new_tokens=200
)
```

```python
tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
outputs = model.generate(
    **inputs,
    max_new_tokens=20,
    do_sample=True, # 추론 결과를 다양하게 생성
    top_p=0.9, # 상위 노출되는 값
    temperature=0.4, # do_sample이 True로 설정 되었을 때 나온 내용들을 자유롭게 생성, 값이 크면 상상력 발휘
    pad_token_id=tokenizer.eos_token_id
)
```

```python
tokenizer.decode(outputs[0], skip_special_tokens=True)
```
